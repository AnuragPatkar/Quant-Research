## Derivative calculations

central difference formula - a numerical method used to estimate the first derivative of a function f(x) at a specific point x using a small step size h.
$$f'(x) \approx \frac{f(x+h) - f(x-h)}{2h}$$


```python

import math

# Exponential function
def exponential(x):
    return math.exp(x)

# Natural logarithm
def natural_log(x):
    return math.log(x)

# Numerical derivative using central difference
def numerical_derivative(func, x, h=1e-5):
    return (func(x + h) - func(x - h)) / (2 * h)

# Test exponential derivative
x = 2

exp_value = exponential(x)
exp_derivative = numerical_derivative(exponential, x)

print("e^x:", exp_value)
print("Derivative of e^x:", exp_derivative)

# Test logarithmic derivative
log_value = natural_log(x)
log_derivative = numerical_derivative(natural_log, x)

print("ln(x):", log_value)
print("Derivative of ln(x):", log_derivative)

# Log return calculation
p0 = 20000
p1 = 22000

simple_return = (p1 - p0) / p0
log_return = math.log(p1 / p0)

print("Simple return:", simple_return)
print("Log return:", log_return)
```

### Numerical integration from scratch: trapezoidal rule

```python
def f(x):
    return 2*x

a = 0
b = 2
n = 100

dx = (b-a)/n
total_aera = 0
for i in range(n):
    x_left = a + i*dx
    x_right = x_left + dx

    y_left = f(x_left)
    y_right = f(x_right)

    aera = ((y_left + y_right) *dx)/2

    total_aera += aera

print(total_aera)
```

# Python — Improper Integral from Scratch

We want to numerically approximate the improper integral:

$$
\int_1^\infty \frac{1}{x^2}\,dx = 1
$$


Since a computer cannot iterate to infinity (\(\infty\)), we approximate the area under the curve by using **increasing truncation points (\(b\))** and the **Midpoint Rule** (dividing the area into \(n\) thin rectangles).

## Quick Intuition (Hinglish)

1. **Width (`dx`):** `1` se `b` tak ke raste ko `n` barabar hisson mein baanta.
2. **Midpoint (`x`):** `1 + i * dx` se har patti ke left corner par pahunche, aur `+ 0.5 * dx` jodkar bilkul **beech (midpoint)** par gaye taaki curve ke bahar aur andar ka hissa balance ho jaye.
3. **Height (`y`):** Midpoint `x` par curve ki oonchai (`1 / x**2`) nikali.
4. **Area (`total`):** Har rectangle ka area (`y * dx`) gullak (`total`) mein jodte gaye. Jaise-jaise `b` bada hota hai, total area **`1`** ke kareeb pahunchta jata hai.


```python
def integral_1_to_b(b,n=100000):
    """
    Approximate integral of 1/x^2
    from x = 1 to x = b using the midpoint rule.
    """
    dx = (b-1)/n
    total = 0.0

    for i in range(n):
        x = 1 + (i + 0.5) * dx
        y = 1/(x**2) 
        area = y * dx # aera of rectangal 
        total += area

    return total

for b in [100,1000,10000]:
    result = integral_1_to_b(b)
    print(f"b = {b:>4}: integral ≈ {result:.6f}")
```

Trigonometric Integrals


$$\int_0^{\pi/2} \sin^3 x \cos^2 x \, dx$$

```python
import numpy as np
from scipy.integrate import quad

# Define the integrand
def f(x):
    return np.sin(x)**3 * np.cos(x)**2

# Numerical integration
result, error = quad(f, 0, np.pi/2)

# Analytical solution
analytical = 2/15

# Difference between numerical and analytical result
difference = abs(result - analytical)

print("Numerical result:", result)
print("Estimated error:", error)
print("Analytical result:", analytical)
print("Absolute difference:", difference)
```
